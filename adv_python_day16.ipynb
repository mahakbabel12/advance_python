# Machine Learning (ML) MOdeling:

```python
'''
Machine learning (ML) modeling is the process of creating a program that can learn from data and make predictions 
or decisions without being explicitly programmed.
Here's a simple explanation of the key steps in ML modeling:

1. Define the Problem
Decide what you want your ML model to do. For example:
Predict house prices (regression).
Classify emails as spam or not spam (classification).
Group customers with similar behavior (clustering).

2. Collect Data
Gather the data the model will learn from. For example:
Sales data, images, text, or sensor readings.
Ensure the data is clean, accurate, and relevant.

3. Prepare the Data
Clean the data: Handle missing values or incorrect data.
Transform the data: Convert it into a format the model can understand (e.g., converting text to numbers).
Split the data:
Training set: Used to train the model.
Testing set: Used to check how well the model works on unseen data.

4. Choose a Model
Pick the type of ML model based on your problem. Common models include:
Linear Regression
Decision Trees
Neural Networks
Support Vector Machines (SVMs)

5. Train the Model
Feed the training data to the model so it can learn patterns and relationships in the data.

6. Evaluate the Model
Test the model on the testing set to see how well it performs. Common metrics include:
Accuracy
Precision
Recall
Mean Squared Error (for regression)

7. Tune the Model
Adjust parameters (hyperparameters) to improve the model’s performance, like:
Learning rate
Number of layers (in a neural network)
Maximum depth (in a decision tree)

8. Deploy the Model
Once the model works well, use it in real-world applications to make predictions.

9. Monitor and Update
Regularly check the model’s performance and retrain it if needed as new data becomes available.
'''
```

# Machine learning can be divided into four main types


```python
#depending on how the model learns and the kind of problem it solves:
'''
1. Supervised Learning
What it does: The model learns from labeled data. Each input has a corresponding correct output, and the goal is to
predict the output for new, unseen inputs.
Examples:
Predicting house prices based on features like size, location, and age (regression).
Classifying emails as spam or not spam (classification).
Algorithms:
Linear Regression
Decision Trees
Support Vector Machines (SVMs)
Neural Networks
                                                                                                                                      
2. Unsupervised Learning
What it does: The model works with unlabeled data and tries to find hidden patterns or structures in the data.
Examples:
Grouping customers with similar purchasing habits (clustering).
Identifying unusual network activity that could indicate a cyberattack (anomaly detection).
Algorithms:
K-Means Clustering
Principal Component Analysis (PCA)
Autoencoders
                  
3. Semi-Supervised Learning
What it does: Combines both labeled and unlabeled data. A small amount of labeled data helps the model learn, while the large amount of unlabeled data improves its performance.
Examples:
Identifying rare diseases where labeled medical data is scarce.
Classifying documents using a mix of labeled and unlabeled text.
Why it’s used: It’s helpful when labeling data is expensive or time-consuming.
                  
4. Reinforcement Learning
What it does: The model learns by interacting with an environment and receiving feedback in the form of rewards or penalties. The goal is to maximize the total reward over time.
Examples:
Teaching robots to walk or play soccer.
Training AI to play games like chess or Go.
Optimizing traffic light systems for smoother traffic flow.
Key Components:
Agent: The learner (e.g., a robot).
Environment: Where the agent interacts (e.g., a game).
Reward: Feedback for the agent’s actions.
'''
```

```python
import pandas as pd
```

```python
df = pd.read_csv("insurance.csv")
```

```python
df.head(2)
```

```python
df.isnull().sum()
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
lb=LabelEncoder()
```

```python
df['sex']=lb.fit_transform(df['sex'])
df['region']=lb.fit_transform(df['region'])
df['smoker']=lb.fit_transform(df['smoker'])
```

```python
df.head(2)
```

```python
x = df.drop(columns = ['charges'])
```

```python
y=df['charges']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr=LinearRegression()
```

```python
lr.fit(x_train , y_train)
```

```python
y_pred = lr.predict(x_test)
```

```python
#It tells you how well the model's predictions align with the actual data.
from sklearn.metrics import r2_score
```

```python
r2_score(y_test ,y_pred)
```
